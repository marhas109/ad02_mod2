### Step 1 ###

Get the file Path for the King County dataset. Save it to a variable using csv_file_path. Make sure you have the path to your data file. For this example, it is in the repository root folder.

```python
csv_file_path = "./King_County_Real_Estate_Sales.csv"
```

### Step 2 ###

Import the csv module.

```python
import csv
```

### Step 3 ###

Load the CSV into a structure where ach row is a dictionary. Store it as a variable data and inspect the first element.

```python
with open(csv_file_path) as f:
    reader = csv.DictReader(f)
    data = list(reader)

data[0]
```

### Step 4 ###

There are a various columns that should have numeric entries.

- 'SalePrice' is quantitative.
- 'PropertyType', 'PrincipalUse', and 'SaleInstrument' are integers encoding categoricals. These are types -- so this is an example of label encoding a nominal categorical.
- 'PropertyClass' is an ordinal encoding the quality and condition of a property.
- 'ExciseTaxNbr' is a unique identifier for a row. 

Convert all columns currently stored as strings to their appropriate data types.

```python
for row in data:
    row['SalePrice'] = int(row['SalePrice'])
    row['PropertyType'] = int(row['PropertyType'])
    row['PrincipalUse'] = int(row['PrincipalUse'])
    row['SaleInstrument'] = int(row['SaleInstrument'])
    row['PropertyClass'] = int(row['PropertyClass'])
    row['ExciseTaxNbr'] = int(row['ExciseTaxNbr'])

data[0]
```

### Step 5 ###

Clean the buyer and seller name strings by removing whitespace.

```python
for row in data:
    row['SellerName'] = row['SellerName'].strip()
    row['BuyerName'] = row['BuyerName'].strip()

data[0]
```

### Step 6 ###

Create new columns for the month, day, and year of the sale transaction and include these in the dataset. Save these into `Month`, `Day`, and `Year` columns. Represent these as numbers of the correct type.

```python
for row in data:
    mmddyy_list = row['DocumentDate'].split('/')
    row['Month'], row['Day'], row['Year'] = int(mmddyy_list[0]), int(mmddyy_list[1]), int(mmddyy_list[2])

data[0]
```

### Step 7 ###

For the column 'AFHistoridProperty', cast Y and N as booleans.

```python
for row in data:
    if row['AFHistoricProperty'] == 'N':
        row['AFHistoricProperty'] = False
    elif row['AFHistoricProperty'] == 'Y':
        row['AFHistoricProperty'] = True

data[0]
```

### Step 8 ###

Create a new version of the dataset with just the columns that have been cleaned and save this to a variable called `cleaned_data`.

```python
col_list = ['ExciseTaxNbr', 'Month', 'Day', 'Year', 'SalePrice', 'PropertyType', 'PropertyClass', 'PrincipalUse', 'SaleInstrument', 'SaleReason', 'AFHistoricProperty']
```

```python
cleaned_data = []

for row in data:
    cleaned_row = {}
    for col in col_list:
        cleaned_row[col] = row.get(col)
    cleaned_data.append(cleaned_row)

cleaned_data[0]
```

### Step 9 ###

Save this dictionary to a new CSV called `cleaned_data.csv` using DictWriter.

```python
with open("cleaned_data.csv", "w") as f:
    csv.DictWriter(f, cleaned_data)
```
