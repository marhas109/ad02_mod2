### Overview ###

A small vet clinic is treating a virus that turns good dogs into bad dogs. There is also a dog flu circulating at the same time.

Data entry for the patients has been rushed and inconsistencies need to be addressed. The data also needs to be in the correct type for upload to the parent company's standards.

Measurements need to be converted to metric, and a new field needs to be added to flag dogs that come in with a fever.

```python
# loading data

dog_data = [['Name', 'Breed', 'Good Doggy?', 'Length [in]', 'Weight Range', 'Body Temp [F]'],
    ["Buddy", "Golden Retriever", "Good", 24, 'Overweight', 101.5],
    ["Cooper", "Doberman Pinscher", "Bad", 28, 'Normal',  105.2],
    ["Max", "GERMAN Shepherd", "Good", 26, 'Normal', 100.2],
    ["Bella", "Chow Chow", "Bad", 20, 'Normal', 100.5],
    ["Lucy", " Yorkshire Terrier", "Good", 9, 'Normal', 102.0],
    ["Charlie", "Beagle", "Good", 13, 'Overweight', 100.4],
    ["Daisy", "Akita", "Bad", 24, 'Normal',99.8],
    ["Bailey", "JACK russell terrier", "Bad", 10, 'Underweight', 101.2],
    ["Lola", "Rottweiler   ", "Good", 27, 'Normal', 102.5],
    ["Sadie", "Siberian_Husky", "Bad", 23, 'Normal', 99.5],
    ["Penny", "STANDARD Poodle", "Good", 18, 'Underweight', 100.7],
    ["Bruno", "Bulldog", "Bad", 14, 'Normal', 104.1],
    ["Rosie", "Beagle", "Good", 13, 'Normal', 100.6],
    ["Duke", "    DOBERMAN PINSCHER", "Good", 28,  'Normal',100.1],
    ["Luna", "Shiba Inu", "Good", 17, 'Obese',  100.3],
    ["Scout", "Golden_Retriever", "Good", 22, 'Normal', 103.5],
    ["Rex", "Rottweiler", "Bad", 23, 'Normal', 100.2],
]
```

#### 1) Write a clean_string function that removes leading/trailing spaces, replaces underscores with spaces, and capitalizes the first letter of each word. Use this function to clean the 'Breed' column. ####

```python
def clean_string(text):
    return text.strip().replace("_", " ").title()

for row in dog_data[1::]:
    row[1] = clean_string(row[1])

dog_data
```

#### 2) Transform lengths from inches to centimeters. ####

```python
# 1 in = 2.54 cm

for i, row in enumerate(dog_data):
    if i > 0:
        row[3] = row[3] * 2.54

dog_data[0][3] = 'Length [cm]'

dog_data
```

#### 3) Convert the Good Doggy? column to a boolean. ####

```python
for i, row in enumerate(dog_data):
    if i > 0:
        row[2] = (row[2] == 'Good') # returns true if good, otherwise false

dog_data
```

#### 4) Convert the Weight column to an integer. ####

```python
value_mapper = {'Underweight': 0, 'Normal': 1, 'Overweight': 2, 'Obese': 3}

for i, row in enumerate(dog_data):
    if i > 0:
        row[4] = value_mapper[row[4]]

dog_data
```

#### 5) Create a new column named Has_Fever. Value should be 0 if temp is normal, 1 if temp is above 102 F. ####

```python
for i, row in enumerate(dog_data):
    if i == 0:
        row.append("Has_Fever")
    else:
        row.append(row[5] > 102.0)

dog_data
```
