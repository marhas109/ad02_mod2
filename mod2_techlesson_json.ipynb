## Overview ##

JSON is not a tabular format with one set of rows and one set of columns. JSON files are often nested in a hierarchical structure and have data structures analogous to Python dictionaries and lists.

Rather than writing code to parse a file's contents, we will use JSON because it is much more efficient. JSON files are ubiquitous in data science, so it is important to learn how to work with them.

This exercise walks through the use of the JSON module to load and extract information from a text-formatted .json file.

1. Loading the data using the JSON file.load() method
2. Using Python data structures to navigate through parsed JSON data
3. Extracting specific data using JSON keys

##### Step 1: JSON Module #####


```python
import json
```

##### Step 2: Load a JSON File #####

```python
with open('nyc_2001_campaign_finance.json') as f: # open with built-in Python open function
    data = json.load(f) # pass the file to json.load

print(type(data))
```

##### Step 3: Parsing a JSON File #####

```python
data.keys()
```

```python
for v in data.values(): 
    print(type(v)) # exploring data types associated with the keys
```

##### Step 4: Parsing Metadata #####

```python
data['meta'].keys() # what are keys of nested dict?
```

```python
type(data['meta']['view']) # what type of value associated with the key?
```

```python
data['meta']['view'].keys() # what are keys of twice-nested dict?
```

##### Step 5: Extracting a Value from a JSON File #####

From the initial exploration, this JSON file contains meta and data. Meta is high-level information and data is the actual campaign finance records.

```python
data['meta']['view'].keys() # looking at the keys of meta again
```

'description' is the 7th key. What's the value associated with its key?

```python
data['meta']['view']['description']
```

This string indicates precisely what the dataset contains.

##### Step 6: Heterogeneous Nested Data Types in JSONs #####

Interestingly, the values insited the nested 'view' dictionary are of mixed type. 

```python
print(data['meta']['view']['numberOfComments'])
print(type(data['meta']['view']['numberOfComments']))
```

Other keys, such as the 'tags' key, are lists of topics the dataset is relevant to.

```python
print(data['meta']['view']['tags'])
print(type(data['meta']['view']['tags']))
```

Other values reveal more structured information in a dictionary, such as the author of the table.

```python
print(data['meta']['view']['tableAuthor'])
print(type(data['meta']['view']['tableAuthor']))
```

Going down another level reveals the author's display name.

```python
print(data['meta']['view']['tableAuthor']['displayName'])
print(type(data['meta']['view']['tableAuthor']['displayName']))
```

Metadata is thus encoded in a logical but non-tabular format. Finding information involves crawling through a hierarchy of data structures, exploring them, then extracting the desired values.
