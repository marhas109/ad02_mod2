### Overview ###

Demonstration of using and manipulating integers, floats, and booleans.

```python
# Exoplanet data
exo_data = [['Name', 'Distance (LY)', 'Orbital Period (Days)', 'Number of Moons', 'Habitable Zone', 'Potential for Liquid Water'],
    ['Kepler-186f', '512', '136', '1', 'TRUE', 'High'],
    ['TRAPPIST-1d', '40', '4.05', '0', 'TRUE', 'High'],
    ['Gliese 581 g', '20', '37', '0', 'FALSE', 'Possible'],
    ['LHS 1140b', '40.5', '38.1', '0', 'FALSE', 'Possible'],
    ['KOI-3140b', '178', '4.5', '0', 'FALSE', 'Low'],
    ['Kepler-1708b', '5600', '131', '2', 'FALSE', 'None'],
    ['Kepler-452b', '1400', '385', '0', 'FALSE', 'Likely'],
    ['Kepler-10b', '186', '0.8', '0', 'FALSE', 'None'],
    ['HD 132563 b', '218', '259.4', '1', 'FALSE', 'Possible'],
    ['Kepler-442b', '1120', '112', '0', 'FALSE', 'Possible']
]
```

### Step 1: Dealing with Integers ###

**Recognizing and converting count data to an integer data type**

The number of moons a planet has must be a whole number and therefore should be stored as an integer.

```python
# loop over rows, using index to skip the first row

for ix, row in enumerate(exo_data):
    if ix > 0:
        row[3] = int(row[3]) 
```

```python
# validating data type of number of moons

type(exo_data[6][3]) is int
```

```python
# confirming all of number of moons are integers (no quotation marks)

exo_data
```

**Recognizing ordinal categoricals and encoding as integers**

The column 'Potential for Liquid Water' is a categorical but not automatically ordered (i.e. numerical)

```python
# extract values of this field w/o column name

lqwater = []

for i, elem in enumerate(exo_data):
    if i > 0:
        lqwater.append(elem[5])

lqwater
```

There is an implied order: none < low < possible < likely < high. 

Implementing this order in base Python can be done by creating a dictionary that

1. defines string to integer mapping
2. loops through rows of exoplanet data
3. replaces strings of the categorical with corresponding integers

```python
value_mapper = {'None': 0, 'Low': 1, 'Possible': 2, 'Likely': 3, 'High': 4}

for i, row in enumerate(exo_data):
    if i > 0:
        row[5] = value_mapper[row[5]] # replaces string with integer value

exo_data
```

### Step 2: Dealing with Continuous Numeric Data ###

Floats are used to represent fractional quantities or variables whose values can be a continuous range.

Distance (LY) and Orbital Period (Days) can both be represented as floats.

```python
for i, row in enumerate(exo_data):
    if i > 0:
        row[1] = float(row[1])
        row[2] = float(row[2])

print("Distance (LY) is a: " + str({type(exo_data[1][1])}))
print("Orbital Period (Days) is a: " +str({type(exo_data[1][2])}))

exo_data
```

It is now possible to perform arithmetic operations on these two numeric columns, such as changing units.

```python
# converting orbital period from days to years

for i, row in enumerate(exo_data):
    if i > 0:
        row[2] = row[2]/365.242374

exo_data[0][2] = 'Orbital Period (Years)' # updating column name

exo_data
```

```python
# validating data type

type(exo_data[1][2])
```

Python also has built in suport for mathematical functions through the mat module:

- math.sqrt
- math.sin
- math.cos
- math.log

Astrophysical distances are often expressed as logs because of the variance in orders of magnitude so we'll convert distance (LY) to log

```python
import math

for i, row in enumerate(exo_data):
    if i > 0:
        row[1] = math.log(row[1])

exo_data[0][1] = 'log(Distance (LY)'

exo_data
```

### Step 3: Creating and Managing Binary Variables ###

**Converting strings to boolean data types**

The habitable zone field is an example of a binary categorical, where each option can be expressed as true or false. Converting this field to boolean datatype will make it easier to manipulate.

```python
for i, row in enumerate(exo_data):
    if i > 0:
        if row[4] == 'TRUE':
            row[4] = True
        elif row[4] == 'FALSE':
            row[4] = False

print(type(exo_data[1][4]))

exo_data
```

Binary data can also be encoded as integers: 0 for false, 1, for true. 

Python implicitly understands this conversion--calling the int() method on a bool returns the integer values.

```python
print(int(True))
print(int(False))
```

Binary categoricals can also be used to segment numeric features, such as recording whether a numeric is above or below a particular threshold.

Below, we bin explanetary data into close and distant.

```python
for i, row in enumerate(exo_data):
    if i == 0:
        row.append('Is Close') # creates new column
    else:
        row.append((row[1] <= 5.0)) # populates boolean entries

exo_data
```
