### Overview ###

Clean the text below by removing:

- erroneous numeric characters
- tab (\t)
- new line special characters (\n)
- leading whitespace

```python
# text to be cleaned

free_text = "\t       Mary had a littl0e lamb and th0en I ate it \n "
print(free_text)
free_text
```

#### Step 1: Whitespace, Inconsistent Case, Unwanted Characters ####

It is customary to standardize all text into lowercase for downstream processing. 

In this step we'll set everything to lowercase and remove unnecessary characters through chained tasks.

```python
cleaned_free_text = free_text.strip().replace('0', '').lower()

cleaned_free_text
```

#### Step 2: Structuring Free Text ####

This cleaned data is still unstructured. The next step is to tokenize (segment) the sentence into semantically useful units. For this text, that means separating the string into a list of constituent words.

The .split(sep) method splits a string on a given separator into a list of strings. The default separator is '\s' (spaces).

```python
token_list = cleaned_free_text.split()

token_list
```

#### Step 3: Reversing a Split (When Applicable) ####

Reversing the splitting process is useful when a cleaned and case-normalized version of the original text is needed. 

The join(iterable) method takes an iterable (tuple, list) and joins all of the elements of the list, separated by a specified string.

```python
clean_restored_free_text = " ".join(token_list)

cleaned_free_text
```
